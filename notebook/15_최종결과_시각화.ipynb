# 15. 최종 결과 시각화

AI 고위험지역 예측과 인프라 사각지대 우선순위를 시각화한다. 결과 PNG는 `results/`에 미리 생성되어 있다.


```python
from pathlib import Path
from IPython.display import Image, display
BASE = Path("..")
for name in [
    "model_comparison_pr_auc.png",
    "2026_high_risk_top15.png",
    "2026_high_risk_map_229.png",
    "infrastructure_blind_spots_top15.png",
    "infrastructure_blind_spots_map_49.png",
]:
    p = BASE / "results" / name
    if p.exists():
        display(Image(filename=str(p)))
```

## 발표용 핵심 메시지

1. 과거 사고·인구·자동차 데이터를 이용해 다음 연도의 상대적 고위험 시군구를 예측했다.
2. 시간 순서 기반 검증으로 미래 정보 누출을 줄였다.
3. 상세 인프라가 확보된 49개 지역에서는 AI 위험도와 시설 부족도를 결합해 현장점검 우선지역을 제안했다.
4. 결과는 인과관계나 자동 정책결정이 아니라 의사결정 보조용 우선순위다.
