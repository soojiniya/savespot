# TAAS 전국 어린이 교통사고 특성 EDA

이 노트북은 `data_processed/taas/charts/`의 차트용 집계 CSV를 사용한다. 자료는 지역명, 학교명, 좌표가 없는 전국 집계 통계이므로 특정 시군구의 특성으로 해석하지 않는다. 사고건수가 많은 시간대는 전국 집계에서 사고가 많이 기록된 시간대라는 뜻이며, 개인별 사고 확률이 높다는 뜻으로 단정하지 않는다.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebook' else Path.cwd()
CHART_DIR = PROJECT_ROOT / 'data_processed' / 'taas' / 'charts'

plt.rcParams['font.family'] = ['Malgun Gothic', 'AppleGothic', 'NanumGothic', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

def load(name):
    return pd.read_csv(CHART_DIR / name, encoding='utf-8-sig')

yearly = load('taas_chart_yearly_totals.csv')
weekday = load('taas_chart_year_weekday.csv')
time = load('taas_chart_year_time.csv')
weekday_time = load('taas_chart_year_weekday_time.csv')
major = load('taas_chart_year_accident_type_major.csv')
car_person = load('taas_chart_year_car_person_middle.csv')
validation = load('taas_chart_aggregation_validation.csv')

display(yearly)
display(validation)
```

## 집계 검증

차트용 집계는 전처리 검증 총계와 모두 일치한다. 연도별 총계, 요일별 합계, 시간대별 합계, 요일×시간대 합계, 사고유형 대분류 합계, 차대사람 중분류 합계가 각각 원본 검증 총계와 맞는지 확인했다.

```python
fig, ax1 = plt.subplots(figsize=(9, 4))
ax1.plot(yearly['연도'], yearly['사고건수'], marker='o', label='사고건수')
ax1.set_ylabel('사고건수')
ax1.set_xlabel('연도')
ax1.grid(axis='y', alpha=0.25)

ax2 = ax1.twinx()
ax2.plot(yearly['연도'], yearly['사망자수'], marker='s', color='crimson', label='사망자수')
ax2.plot(yearly['연도'], yearly['부상자수'], marker='^', color='darkorange', label='부상자수')
ax2.set_ylabel('사상자수')

lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines + lines2, labels + labels2, loc='upper left')
ax1.set_title('연도별 전국 어린이 교통사고 집계')
plt.tight_layout()
```

관찰 결과: 사고건수는 2020년 8,400건에서 2022년 9,163건으로 증가했고, 2023년 8,753건으로 내려간 뒤 2024년 9,149건, 2025년 8,780건으로 집계됐다. 사망자수는 2020년 24명에서 2025년 13명으로 낮아졌다. 부상자수는 2022년 11,389명이 가장 많고 2020년 10,500명이 가장 적다.

```python
fig, ax = plt.subplots(figsize=(10, 4))
for day, part in weekday.sort_values(['요일 정렬값', '연도']).groupby('요일'):
    ax.plot(part['연도'], part['연도내_사고건수비율'] * 100, marker='o', label=day)
ax.set_title('연도별 요일 사고건수 비율')
ax.set_xlabel('연도')
ax.set_ylabel('연도 전체 사고건수 대비 비율(%)')
ax.grid(axis='y', alpha=0.25)
ax.legend(ncol=7, loc='upper center', bbox_to_anchor=(0.5, -0.16))
plt.tight_layout()
```

관찰 결과: 2020~2025년 모두 토요일 사고건수가 가장 많다. 토요일 사고건수는 2020년 1,494건, 2024년 1,730건, 2025년 1,518건이며 연도별 비율은 약 17.2~18.9% 범위다.

```python
fig, ax = plt.subplots(figsize=(11, 4.5))
for year, part in time.groupby('연도'):
    part = part.sort_values('시작시각')
    ax.plot(part['시간대'], part['연도내_사고건수비율'] * 100, marker='o', label=str(year))
ax.set_title('연도별 시간대 사고건수 비율')
ax.set_xlabel('시간대')
ax.set_ylabel('연도 전체 사고건수 대비 비율(%)')
ax.tick_params(axis='x', rotation=45)
ax.grid(axis='y', alpha=0.25)
ax.legend(title='연도', ncol=6, loc='upper center', bbox_to_anchor=(0.5, -0.28))
plt.tight_layout()
```

관찰 결과: 2020~2025년 모두 16시~18시 사고건수가 가장 많다. 해당 시간대 사고건수는 2020년 1,860건, 2022년 2,182건, 2025년 2,038건이며 연도별 비율은 약 22.1~23.8%다. 이는 전국 집계의 발생 분포이며, 개인의 시간대별 사고 확률을 직접 뜻하지 않는다.

```python
selected_year = 2025
heat = weekday_time[weekday_time['연도'] == selected_year].pivot(index='요일', columns='시간대', values='사고건수')
weekday_order = weekday.sort_values('요일 정렬값')['요일'].drop_duplicates().tolist()
time_order = time.sort_values('시작시각')['시간대'].drop_duplicates().tolist()
heat = heat.reindex(index=weekday_order, columns=time_order)

fig, ax = plt.subplots(figsize=(11, 4.8))
image = ax.imshow(heat.values, cmap='YlOrRd')
ax.set_xticks(range(len(time_order)), time_order, rotation=45, ha='right')
ax.set_yticks(range(len(weekday_order)), weekday_order)
ax.set_title(f'{selected_year}년 요일 x 시간대 사고건수')
for i, day in enumerate(weekday_order):
    for j, band in enumerate(time_order):
        ax.text(j, i, int(heat.loc[day, band]), ha='center', va='center', fontsize=8)
fig.colorbar(image, ax=ax, label='사고건수')
plt.tight_layout()
```

관찰 결과: 2025년 기준 요일×시간대 조합에서는 토요일 14시~16시 284건, 토요일 12시~14시 276건, 수요일 16시~18시 299건, 금요일 16시~18시 332건처럼 오후 시간대에 큰 값이 보인다. 이는 전국 집계상 사고건수의 분포를 보여주는 것으로, 특정 지역의 교통환경이나 개인 노출량을 설명하지 않는다.

```python
pivot_major = major.pivot(index='연도', columns='사고유형 대분류', values='연도내_사고건수비율') * 100
ax = pivot_major.plot(kind='bar', stacked=True, figsize=(10, 4.5), colormap='tab20')
ax.set_title('연도별 사고유형 대분류 사고건수 비율')
ax.set_xlabel('연도')
ax.set_ylabel('연도 전체 사고건수 대비 비율(%)')
ax.legend(title='사고유형 대분류', loc='upper center', bbox_to_anchor=(0.5, -0.18), ncol=4)
plt.tight_layout()
```

관찰 결과: 사고유형 대분류에서는 모든 연도에서 차대차 사고가 가장 많다. 차대차 사고건수는 2020년 6,232건, 2025년 6,320건이며 연도별 전체 사고건수 대비 비율은 약 68.5~74.2%다. 차대사람 사고는 별도 보행 사고 전체와 같은 말이 아니며, 원본 사고유형 대분류의 하나로 해석해야 한다.

```python
pivot_cp = car_person.pivot(index='연도', columns='사고유형 중분류', values='연도별_차대사람내_사고건수비율') * 100
ax = pivot_cp.plot(kind='bar', stacked=True, figsize=(10, 4.5), colormap='Set2')
ax.set_title('연도별 차대사람 사고 중분류 비율')
ax.set_xlabel('연도')
ax.set_ylabel('차대사람 사고건수 대비 비율(%)')
ax.legend(title='사고유형 중분류', loc='upper center', bbox_to_anchor=(0.5, -0.18), ncol=3)
plt.tight_layout()
```

관찰 결과: 차대사람 사고 중분류에서는 모든 연도에서 횡단중이 가장 많다. 횡단중 사고건수는 2020년 976건, 2022년 1,352건, 2025년 1,275건이며 차대사람 사고 내 비율은 약 46.9~54.0%다. 2025년 사고유형 계층은 2020~2024년과 일부 다르므로, 원본에 없는 유형을 0으로 채워 비교하지 않았다.

## 해석 주의

- 이 노트북은 전국 사고 특성 분석용이다. 시군구별 예측 결과에 전국 수치를 복제해 붙이지 않는다.
- 사고건수와 사망자수·부상자수는 단위가 다르므로 합산하지 않는다.
